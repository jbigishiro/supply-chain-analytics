```python
from statsmodels.stats.proportion import proportions_ztest, confint_proportions_2indep

# returned customers and group sizes, from sql/analysis/15_repeat_purchases.sql
returned  = [1030, 18 + 41]          # on time, all late
customers = [51558, 987 + 2682]

z, p = proportions_ztest(returned, customers)
low, high = confint_proportions_2indep(returned[0], customers[0], returned[1], customers[1])

print(f"z = {z:.2f}, p = {p:.3f}")
print(f"95% CI for the difference: {low*100:.2f} to {high*100:.2f} percentage points")
```
