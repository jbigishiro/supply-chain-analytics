```python
import math
import pandas as pd
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

baseline = 0.524        # 1–2 star rate among triggered orders (query 17)
review_rate = 0.983     # share of triggered orders that leave a review
weekly = 176            # triggered orders per week (Jan–Jun 2018)

rows = []
for mde in [0.03, 0.05, 0.075, 0.10]:                       # absolute drop, percentage points
    h = proportion_effectsize(baseline, baseline - mde)       # Cohen's h
    n_reviews = NormalIndPower().solve_power(h, alpha=0.05, power=0.8, ratio=1)
    n_orders = math.ceil(n_reviews / review_rate)             # inflate: not every order is reviewed
    rows.append({"mde_pp": mde * 100,
                 "treated_rate": baseline - mde,
                 "orders_per_arm": n_orders,
                 "total_orders": 2 * n_orders,
                 "weeks": round(2 * n_orders / weekly, 1)})
pd.DataFrame(rows)
```
